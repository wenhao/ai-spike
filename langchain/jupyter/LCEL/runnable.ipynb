```python
from langchain_openai import ChatOpenAI
import os

model = ChatOpenAI(
    model_name="qwen-max",
    openai_api_key = os.getenv("DASHSCOPE_API_KEY"),
    openai_api_base = "https://dashscope.aliyuncs.com/compatible-mode/v1",
    temperature=0.9, 
    max_tokens=5000,
)
```

```python
from langchain_core.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_template("讲个关于 {topic} 的笑话吧")
chain = prompt | model
```

```python
chain.invoke("程序员")
```

```python
for chunk in chain.stream({"topic": "程序员"}):
    print(chunk.content, end="", flush=True)
```

```python
chain.batch([{"topic": "程序员"}, {"topic": "产品经理"}, {"topic": "测试经理"}])
```

```python
from langchain_core.output_parsers import StrOutputParser

output_parser = StrOutputParser()

messages = chain.batch([{"topic": "程序员"}, {"topic": "产品经理"}, {"topic": "测试经理"}])

for idx, message in enumerate(messages):
    print(f"笑话{idx}:\n")
    print(output_parser.invoke(message))
    print("\n")
```

```python
async for s in chain.astream({"topic": "程序员"}):
    print(s.content, end="", flush=True)
```
